```python
v = "otro texto"
n= 10
print("Un texto",v,"y un numero",n)
```

```python
c = "Un texto '{}' y un numero '{}'".format(v,n)
c
```

```python
print(c)
```

```python
##Donde 0 y 1 representan la posicion como si fuera un arreglo
print("Un texto '{1}' y un numero '{0}'".format(v,n) )
```

```python
print("Un texto '{texto}' y un numero '{numero}'".format(texto= v, numero = n) )
```

```python
print("Un texto '{v}' y un numero '{n}'".format(v= v, n = n) )
```

```python
print("{v},{v},{v}".format(v=v))
```

```python
###ALINEAMIENTO A LA DERECHA
print("{:>30}".format("palabra"))
```

```python
###ALINEAMIENTO A LA IZQUIERDA
print("{:30}".format("palabra"))
##selecciona el resultado y hay 3 caracteres a la derecha
```

```python
###ALINEAMIENTO A CENTRO
print("{:^30}".format("palabra"))
```

```python
###TRUNCAMIENTO A 3 CARACTERES
print("{:.3}".format("palabra"))
```

```python
###ALINEAMIENTO A LA DERECHA CON TRUNCAMIENTO
print("{:>30.3}".format("palabra"))
```

```python
##numeros enteros rellenados con espacio
print("{:4d}".format(10))
print("{:4d}".format(100))
print("{:4d}".format(1000))
```

```python
##numeros enteros rellenados con 0
print("{:04d}".format(10))
print("{:04d}".format(100))
print("{:04d}".format(1000))
```

```python
##formateo de numero flotantes con espacio
print("{:7.3f}".format(3.14567))
print("{:7.3f}".format(153.21))
```

```python
##formateo de numero flotantes con 0
print("{:07.3f}".format(3.14567))
print("{:07.3f}".format(153.21))
```
